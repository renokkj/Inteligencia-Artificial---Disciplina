4. Gere um array de 100 números inteiros aleatórios e encontre a mediana e
a moda dos valores.


```python
import numpy as np;

# gera os inteiros aleatorios
array = np.random.randint(0, 101, 100);

# encontra o valor central para mediana
mediana = np.median(array);
# extrai os valores unicos
valor = np.unique(array);

cont = []

for v in valor:
  # conta quantas vezes o valor aparece
  qtd = (array == v).sum();
  cont.append(qtd);

#converte para usar as funçoes de array numpy
cont = np.array(cont);
# np.argmax = devolve o indice do valor maximo no array
ind = np.argmax(cont);
moda = valor[ind];

print(f'Array: {array}');
print();
print(f'Mediana: {mediana}');
print(f'Moda: {moda}');
print(f'Aparece: {cont[ind]}');
```



*   **np.median()** = Função para encontrar o valor central (após ordear os dados);
*   **np.unique()** = Função para encontrar moda (identificar valores diferentes);
*   **np.argmax()** = Função para localizar o maior índice de frequência com lógica de conjuntos

7. Crie dois arrays de números aleatórios e encontre a correlação entre eles.

```python
import numpy as np;

# arrays de numeros aleatorios de 0 a 1
array1 = np.random.rand(10);
array2 = np.random.rand(10);

media1 = np.mean(array1);
media2 = np.mean(array2);

# np.mean = media // np.std = desvio padrao
num = np.mean((array1 - media1) * (array2 - media2));
den = np.std(array1) * np.std(array2);

# correlaçao = media do produto dos devios / produto dos desvios padrao
co = num / den;

print('Array 1: ');
print(array1);
print();
print('Array 2: ');
print(array2);
print();
print('Correlação: ');
print(co);
```



*   **np.mean()** = Calcula a média aritmética dos valores do array
*   **np.std()** = Calcula o desvio padrão, mede o quanto os dados variam em relação à média geral;

19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o
ângulo entre eles (produto escalar normalizado).


```python
import numpy as np;

# arrays de numero aleatorio ate 20 caracteres
array1 = np.random.randint(1, 101, 20);
array2 = np.random.randint(1, 101, 20);
# calculo produto escalar
prod = np.dot(array1, array2);
# calculo da magnitude (mag = raiz dos elementos ao quadrado)
mag1 = np.sqrt(np.sum(array1 ** 2));
mag2 = np.sqrt(np.sum(array2 ** 2));
# calculo produto escalar normalizado
cos = prod / (mag1 * mag2);

print('Array 1:');
print(array1);
print();
print('Array 2:');
print(array2);
print();
print(f'Produto escalar: {prod}');
print();
print(f'Produto escalar normalizado: {cos}');
```

4. Use a função de otimização de SciPy para maximizar f(x) = −x²+ 4x+ 1.

```python
import numpy as np;
from scipy.optimize import minimize;

# funçao original
def f(x):
  return -x ** 2 + 4 * x +1;
# maximizaçao
def fneg(x):
  return -f(x);
# otiizaçao com valor inicial 0
r = minimize(fneg, 0);

print(f'Ponto máximo (x): {r.x}');
print(f'Valor máximo f(x): {f(r.x)}');
```

Minimize busca o valor, para maximizar a função retorna o valor negativo de f(x). Minimizar o negativo maximiza a original.

```python
import numpy as np;
from scipy.integrate import solve_ivp;

# definiçao da equaçao diferencial
def edo(x, u):
  # converte segunda ordem para um sistema de primeira
  return [u[1], -u[0]];

u0 = [1, 0]; # y(0) = 1 e y'(0) = 0
span = (0, 10); # intervalo de integraçao
eval = np.linspace(0, 10, 100); # pontos de avaliaçao
# execuçao do solver para valor inicial
solu = solve_ivp(edo, span, u0, t_eval = eval);
# ponto para testar precisao
alv = np.pi;
# encontra a proximidade
dist = np.abs(solu.t - alv);
# ponto mais proximo de pi
prox = np.argmin(dist);
# extrai o valor de y, primeira linha da matriz
calc = solu.y[0, prox];

print(f'Ponto de teste: {solu.t[prox]:.4f}');
print(f'Valor numérico (y): {calc:.4f}');
print(f'Valor teórico(cos(pi)): {np.cos(alv)}');
```

Busca do ponto foi dividida em variáveis para melhor entendimento das operações;

np.argmin() = Localiza a posição do menor valor;

solve_ivp() = Resolve problemas de valor inicial,

solu.t[prox] = Acesso a primeira variável de estado (y) dentro da matriz;

plt.subplot(2, 1, n) = Cria uma grade de linhas e colunas para comparar as diferentes visualizações de dados;

plt.tight_layout() = Ajusta automaticamente o layout dos gráficos;

8. Plote um gráfico de linha com intervalos de confiança (sombramento) para
a função y = sin(x).


```python
import matplotlib.pyplot as plt;
import numpy as np;

# gera os dados base
x = np.linspace(0, 10, 100);
y = np.sin(x);

# intervalo de confiança
erro = 0.2; # sombreamento
sup = y + erro;
inf = y - erro;

# plotagem
plt.plot(x, y, color = 'blue', label = 'y = sen(x)');

# adiçao do sombreamento
plt.fill_between(x, inf, sup, color = 'blue', alpha = 0.2, label = 'Intervalo de confiança');

plt.title('Gráfico de seno');
plt.xlabel('Eixo X');
plt.ylabel('Eixo Y');
plt.legend();
plt.show();
```

plt.fill_between(x, y1, y2) = Visualizar a variabilidade de um modelo, 'sombreamento';

alpha= 0.2 = Permite que a curva principal e a grade do gráfico permaneçam visíveis sob a área sombreada;

18. Crie um gráfico de barras com valores positivos e negativos e use diferentes
cores para cada tipo de valor.


```python
import matplotlib.pyplot as plt;
import numpy as np;

cat = ['Categoria A', 'Categoria B', 'Categoria C', 'Categoria D'];
val = [12, -7, 18, -10];

# cores por polaridade
cor = [];
for v in val:
  if v > 0:
    cor.append('green');
  else:
    cor.append('red');

# grafico de barras
plt.bar(cat, val, color = cor, edgecolor = 'black');

# metadados
plt.title('Variação de categorias');
plt.xlabel('Categorias analisadas');
plt.ylabel('Variação de valor');

# linha de referencia para polaridade
plt.axhline(0, color = 'black', linewidth = 0.8);

plt.show();
```

plt.bar() = Compara diferentes categorias ou grupos;

plt.axhline() = Usa linhas como referência para melhor visualização e legibilidade no eixo 0;

```python
import matplotlib.pyplot as plt;
import numpy as np;

x = np.linspace(0, 10, 100);
y1 = np.sin(x); # seno (primeira serie)
y2 = np.cos(x); # cos (segunda serie)

# grafico com rotulos
plt.plot(x, y1, label = 'Seno', color = 'blue');
plt.plot(x, y2, label = 'Cosseno', color = 'red', linestyle = '--');

# metadados
plt.title('Grafico de linha com comparação');
plt.xlabel('Eixo X');
plt.ylabel('Eixo Y');
plt.legend(loc = 'best'); # escolhe o local com menos sobreposiçao
plt.show();
```

20. Adicione uma grade com espaçamento personalizado a um gráfico de linha
para y = x².

```python
import matplotlib.pyplot as plt;
import numpy as np;

x = np.linspace(-10, 10, 100);
y = x ** 2;

# grafico de linha
plt.plot(x, y, color = 'blue', label = 'y = x²');

# espaçamento
plt.xticks(np.arange(-10, 11, 2)); # grade de 2 em 2 no eixo X
plt.yticks(np.arange(0, 101, 10)); # grade de 10 em 10 no eixo Y

# grade
plt.grid(True, linestyle = ':', alpha = 0.6, color = 'gray');

# metadados
plt.title('Gráfico com grade com espaçamento');
plt.xlabel('Eixo X');
plt.ylabel('Eixo Y');
plt.legend();
plt.show();
```

plt.xticks() / plt.yticks() = Controlam as localizações dos tiques no gráfico, como a grade é desenhada sobre essas marcas é melhor defini-las manualmente (espaçamento personalizado);

np.arange() = Gera sequências de números com passoas específicos;
