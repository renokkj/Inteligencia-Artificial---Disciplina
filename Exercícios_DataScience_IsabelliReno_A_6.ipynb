4. Gere um array de 100 números inteiros aleatórios e encontre a mediana e
a moda dos valores.


```python
import numpy as np;

# gera os inteiros aleatorios
array = np.random.randint(0, 101, 100);

# encontra o valor central para mediana
mediana = np.median(array);
# extrai os valores unicos
valor = np.unique(array);

cont = []

for v in valor:
  # conta quantas vezes o valor aparece
  qtd = (array == v).sum();
  cont.append(qtd);

#converte para usar as funçoes de array numpy
cont = np.array(cont);
# np.argmax = devolve o indice do valor maximo no array
ind = np.argmax(cont);
moda = valor[ind];

print(f'Array: {array}');
print();
print(f'Mediana: {mediana}');
print(f'Moda: {moda}');
print(f'Aparece: {cont[ind]}');
```



*   **np.median()** = Função para encontrar o valor central (após ordear os dados);
*   **np.unique()** = Função para encontrar moda (identificar valores diferentes);
*   **np.argmax()** = Função para localizar o maior índice de frequência com lógica de conjuntos



5. Crie um array de 20 elementos contendo números entre 1 e 100 e substitua
todos os valores múltiplos de 3 por -3.

```python
import numpy as np;

# gera o array original
array = np.random.randint(1, 101, 20);
# cria uma copia do array onde nao é afetado pela conta
arraycop = array.copy();
# substituiçao dos multiplos de 3
arraycop[arraycop % 3 == 0] = - 3;

print('Array original: ');
print(array);
print();
print('Array substituida: ');
print(arraycop);
```

7. Crie dois arrays de números aleatórios e encontre a correlação entre eles.

```python
import numpy as np;

# arrays de numeros aleatorios de 0 a 1
array1 = np.random.rand(10);
array2 = np.random.rand(10);

media1 = np.mean(array1);
media2 = np.mean(array2);

# np.mean = media // np.std = desvio padrao
num = np.mean((array1 - media1) * (array2 - media2));
den = np.std(array1) * np.std(array2);

# correlaçao = media do produto dos devios / produto dos desvios padrao
co = num / den;

print('Array 1: ');
print(array1);
print();
print('Array 2: ');
print(array2);
print();
print('Correlação: ');
print(co);
```



*   **np.mean()** = Calcula a média aritmética dos valores do array
*   **np.std()** = Calcula o desvio padrão, mede o quanto os dados variam em relação à média geral;

9. Crie um array com 10 números aleatórios e normalize seus valores entre 0
e 1.

```python
import numpy as np;

# array com 10 numeros aleatorios entre 0 e 1
array1 = np.random.rand(10);

# valores minimo e maximo para aplicaçao na formula
min = np.min(array1);
max = np.max(array1);

# formula de normalizaçao
array2 = (array1 - min) / (max - min);

print('Array original:');
print(array1);
print();
print('Array normalizado:');
print(array2);
```

10. Crie uma matriz identidade 6x6 e substitua a diagonal principal por valores
crescentes de 1 a 6.


```python
import numpy as np;

# cria a matriz bidimensional 6x6
matriz = np.eye(6);

print('Matriz original:');
print(matriz);

# substitui a diagonal principal pelo valores crescentes
for i in range(6):
  # acessa os elementos especificos da array
  matriz[i, i] = i + i;

print('Matriz substituída:');
print(matriz);
```



*   **negnp.eye()** = Função de criação de matrizes de identidade;
*   **[i, i]** = Acesso aos elementos específicos por sintaxe;
*   **for i in range()** = Manipulação para o preenchimento de diagonal;

11. Gere um array de 100 números inteiros e encontre o valor que mais se
repete no array.

```python
import numpy as np;

array = np.random.randint(0, 51, 100);
# identifica os valores unicos no array
unicos = np.unique(array);
# lista vazia para as contagens
liscont = [];

for valor in unicos:
  # gera eum boolean e conta os True
  qnt = (array == valor).sum();
  liscont.append(qnt);

# converte a lista de volta para um array
cont = np.array(liscont);
# localiza a posiçao valor maximo e o indice de maior frequencia
max = np.argmax(cont);

# resultados
moda = unicos[max];
rep = cont[max];

print('Array original:');
print(array);
print();
print('Valor que mais se repete:');
print(moda);
print();
print('Quantidade de repetições:');
print(rep);
```



*   **np.unique()** = Função que encontra valores únicos na array;
*   **(array == valor)** = Compara cada elemento do array com o valor;
*   **sum()** = Soma os valores da array resultante;

13. Crie um array contendo os 10 primeiros números pares e calcule sua soma
cumulativa.


```python
import numpy as np;

# array com 10 numeros pares
array = np.linspace(2, 20, 10);
# array de zeros para armazenar os resultados
soma = np.zeros(10);

# acumula os valores com fatiamento
for i in range(10):
  # soma do inicio ate a posiçao i + 1
  soma[i] = np.sum(array[: i + 1]);

print('Array 10 números pares:')
print(array);
print();
print('Soma acumulativa:');
print(soma);
```


*   **np.sum(array[: i + 1])** = Seleciona e soma os elementos do array desde o primeiro item até a posição i;
*   Laço **for** permite o processa a soma;
*   **np.zeros()** = Cria o array de zeros para armazenamento dos resultados;
*   **np.linspace()** = Cria a lista de números com espaçamento igual entre eles;

17. Gere uma matriz 4x4 de números inteiros aleatórios e multiplique cada
valor por sua posição na matriz.

```python
import numpy as np;

# gera a matriz aleatoria 4x4
matriz = np.random.randint(1, 11, (4, 4));
# vetor de indices
ind = np.linspace(0, 3, 4);
# transforma em coluna e linha
coluna = np.reshape(ind, (4, 1));
linha = np.reshape(ind, (1, 4));
# gera a multiplicaçao
posi = np.dot(coluna, linha);
# multiplica o valor pelo indice (posicao)
r = matriz * posi;

print('Matriz original:');
print(matriz);
print();
print('Matriz multiplicada:');
print(r);
```



*   **np.linspace()** = Cria a sequência numérica;
*   **np.reshape()** = Muda o formato/dimensões para organizar os dados;

*   **np.dot()** = Calcula o produto escalar de vetores e faz a multiplicação de matrizes;
