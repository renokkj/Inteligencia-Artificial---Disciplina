20. Matriz Transposta - Dada uma matriz (lista de listas), crie sua matriz transposta.


```python
# matriz transposta = troca linhas por colunas

m = [[1, 3, 5], [2, 4, 6]]; # matriz

l = len(m); # linhas
c = len(m[0]); # colunas

t = []; # guarda a matriz transposta
ic = 0; # indice da coluna

while ic < c: # colunas
  nl = []; # nova linha da matriz
  il = 0 # indice da linha
  while il < l: # linhas
    nl.append(m[il][ic]); # pega o elemento i linha e i coluna e coloca em nl
    il = il + 1;
  t.append(nl);
  ic = ic + 1;

print(f'Matriz original: {m}');
print();
print(f'Matriz transposta: {t}');
```

22. Criptografia de Cifra de César - Crie uma função para criptografar uma mensagem usando a cifra de César.


```python
# cifra de cesar = desloca letras num numero de posiçoes

def cc(msg, desloc): # string do alfabeto para deslocar manualmente
  alf = "abcdefghijklmnopqrstuvwxyz";
  r = "";

  msg = msg.lower();

  for letra in msg:

    if letra in alf:
      pos = alf.find(letra); # encontra a posição atual da letra (0 - 25)
      novpos = (pos + desloc) % 26; # calcula a nova posição e da a volta no alfabeto se passar de z
      r += alf[novpos]; # adiciona a letra na nova posição

    else:
      r += letra; # se nao for uma letra mantem como esta

  return r;

t = input(('Frase/palavra para criptografar: '));
d = int(input('Posições para deslocamento: '));

msgcc = cc(t, d); # pega o que o usuario digitou e criptografa

print();
print(f'Mensagem original: {t}');
print(f'Mensagem criptografada: {msgcc}');
```

23. Descriptografar Cifra de César - Crie uma função para descriptografar uma mensagem cifrada com a cifra de César.


```python
# estou usando a mesma base de codigo do exercicio anterior

def dcc(msg, desloc): # string do alfabeto para deslocar manualmente
  alf = "abcdefghijklmnopqrstuvwxyz";
  r = "";

  msg = msg.lower();

  for letra in msg:

    if letra in alf:
      pos = alf.find(letra); # encontra a posição atual da letra
      novpos = (pos - desloc) % 26;
      r += alf[novpos];

    else:
      r += letra;

  return r;

tcc = input('Mensagem criptografada: '); # texto criptografado
dc = int(input('Posições para deslocamento: '));

msgog = dcc(tcc, dc); # mensagem original descriptografada

print();
print(f'Mensagem criptografada: {tcc}');
print(f'Mensagem descriptografada: {msgog}');
```

24. Validação de CPF - Implemente uma função para validar um CPF brasileiro.


```python
# validaçao pelos dois ultimos digitos do cpf

def validar(cpf): # apenas o numero da entrada
  base = "";
  for carac in cpf:
    if carac.isdigit(): # verifica se é um digito
      base += carac;

  if len(base) != 11: # verifica o tamanho
    return False;

  if base == base[0] * 11: # compara o cpf todo com o primeiro digito 11 vezes
    return False;

  soma = 0; # calculo primeiro digito
  for i in range(9):
    soma += int(base[i]) * (10 - i); # multiplica os 9 digitos de 10 a 2
  resto = (soma * 10) % 11;

  d1 = 0 if resto >= 10 else resto; # se o resultado der 10 ou 11, digito = 0
  if d1  != int(base[9]): # compara o calculo com o decimo numero
    return False;

  soma = 0; # calculo segundo numero
  for i in range(10):
    soma += int(base[i]) * (11 - i); # multiplica os 10 digitos de 11 a 2
  resto = (soma * 10) % 11;

  d2 = 0 if resto >= 10 else resto;
  if d2 != int(base[10]): # compara o calculo com o numero real do cpf
    return False;

  return True;

e = input('Digite um CPF: ');
if validar(e):
  print('CPF válido');
else:
  print('CPF inválido');
```

31. Ordenação por Bolha - Implemente o algoritmo de ordenação por bolha (Bubble Sort).


```python
# Bubble Sort = percorre a lista varias vezes // esquerda > direita = troca de lugar (se nao for, passa para o proximo)

num = [3, 6, 8, 1, 4, 2, 5, 7];
n = len(num); # tamanho da lista
lista1 = num[:]; # fatiamento para diferenciar

for i in range(n): # controle passagens
  for j in range(0, n - i - 1): # compara pares e o maior vai para o final

    if num[j] > num[j + 1]: # compara com o do lado
      num[j], num[j + 1] = num[j + 1], num[j]; # troca de lugar

print(f'Lista original: {lista1}');
print(f'Lista ordenada: {num}');
```

36. Gerador de Senhas - Crie um gerador de senhas aleatórias com letras e números.

```python
# aleatoriedade com LCG

let = 'abcdefghijklmnopqrstuvwxyz';
num = '0123456789';
junta = let + num; # junta letras com numeros em uma string

tamanho = int(input('Tamanho da senha (número): '));
nb = int(input('Número base: ')); # pede um numero base para aleatoriedade

def senha(tam, nb):
  sen = "";
  x = nb;

  for i in range(tam):
    x = (1103515245 * x + 12345) % 2 ** 31 # formula de aleatoriedade pela LCG
    ind = x % len(junta); # divide o resto pelo tamanho da string de caracteres
    sen += junta[ind]; # caractere somado a senha

  return sen;

r = senha(tamanho, nb);
print();
print(f'Senha gerada: {r}');
```

37. Funções Recursivas - Implemente uma função recursiva para calcular o fatorial de um número.


```python
# fatorial = produto dos numeros inteiros de 1 ate n

def fat(n):

  if n == 0: # fatorial de 0 sempre é 1
    return 1;
  else: # formula n * (n - 1)!
    return n * fat(n - 1);

print('CALCULO DE FATORIAL');
print();

n = int(input('Digite um número: '));

if n < 0: # validação de numeros negativos
  print('Número inválido');
else:
  r = fat(n);
  print(f'Fatorial de {n}: {r}');
```

40. Lista de Compras - Crie um programa para gerenciar uma lista de compras, permitindo adicionar, remover e listar itens.

42. Números Primos até N - Gere todos os números primos até um número n fornecido.


```python
print('GERAR NÚMEROS PRIMOS ATÉ N')

print();
n = int(input('Valor máximo (n): '));

primos = []; # lista de numeros primos vazia

for num in range(2, n + 1): # começa no 2 pois 0 e 1 nao sao considerados primos
  primo = True;

  for i in range(2, num): # ve se o num é divisivel por 2 ou ele mesmo
    if num % i == 0:
      primo = False;
      break; # nao é primo = encerra

  if primo:
    primos.append(num); # se for primo adiciona na lista

print();
print(f'Números primos até {n}: ');
print(primos);
```

44. Fibonacci em Nível - Gere a sequência de Fibonacci até um determinado número de termos.


```python
# sequencia de termos independende do valor que atinge

nt = int(input('Quantidade de termos: ')); # nt = numero de termos

a, b = 0, 1; # valores padrao da sequencia
seq = [];

for i in range (nt):
  seq.append(a); # adiciona o termo na lista

  a, b = b, a + b; # atualiza os valores

print();
print(f'Primeiros {nt} termos: ');
print(seq);
```

46. Calcular Média e Desvio Padrão - Calcule a média e o desvio padrão de uma lista de números.


```python
lista = [1, 2, 3, 4, 5];
n = len(lista);

media = sum(lista) / n; # calculo de media
quades = 0;

for x in lista: # soma dos quadrados dos desvios em relaçao a media
  desvio = x - media;
  quades += desvio ** 2;

varia = quades / (n -1); # divide a soma por n - 1
desviop = varia ** 0.5; # ** 0.5 = calculo de raiz

print(f'Lista original: {lista}');
print(f'Média: {media:.2f}');
print(f'Desvio padrão: {desviop:.2f}');
```

48. Ordenação por Inserção - Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
lista = [2, 4, 5, 1, 6, 7, 3];
lista2 = lista[:]; # copia da primeira lista

for i in range(1, len(lista)): #começa do segundo elemento, o primeiro é base
  b2 = lista[i]; # segunda base, numero para encaixar
  j = i - 1; # indice da esquerda

  while j >= 0 and lista[j] > b2: # fora do inicio da lista e o da esquerda for maior que a base
    lista[j + 1] = lista[j]; # numero maior vai pra direita
    j -= 1;

  lista[j + 1] = b2; # acha onde a segunda base era pra estar

print(f'Lista original: {lista2}');
print(f'Lista ordenada: {lista}');
```

50. Cálculo de Frequência de Letras - Calcule a frequência de cada letra em uma string fornecida pelo usuário.

```python
t = input('Digite uma frase: ').lower();
freq = {}; # dicionario vazio

for letra in t:
  if letra.isalpha(): # verifica se o caractere é uma letra
    if letra in freq: # se ja tiver no dicionario, soma 1
      freq[letra] += 1;
    else:
      freq[letra] = 1; # se nao tiver no dicionario, começa com 1

print();
print('FREQUENCIA DE LETRAS');
print();

for let in sorted(freq.keys()): # ordena
  print(f'Letra {let}: {freq[let]} vez(es)');
```
