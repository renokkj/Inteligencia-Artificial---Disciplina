9. Fatorial de um Número - Calcule o fatorial de um número fornecido pelo usuário.


```python
n = int(input('Digite um número: '));

x = 1;
f = 1;

while x <= n:
  f = f * x; # f multiplica por x acumulando valor
  x = x + 1; # atualiza para nao entrar em loop infinito

print(f'fat({n}) = {f}');
```

20. Matriz Transposta - Dada uma matriz (lista de listas), crie sua matriz transposta.


```python
# matriz transposta = troca linhas por colunas

m = [[1, 3, 5], [2, 4, 6]]; # matriz

l = len(m); # linhas
c = len(m[0]); # colunas

t = []; # guarda a matriz transposta
ic = 0; # indice da coluna

while ic < c: # colunas
  nl = []; # nova linha da matriz
  il = 0 # indice da linha
  while il < l: # linhas
    nl.append(m[il][ic]); # pega o elemento i linha e i coluna e coloca em nl
    il = il + 1;
  t.append(nl);
  ic = ic + 1;

print(f'Matriz original: {m}');
print();
print(f'Matriz transposta: {t}');
```

21. Busca Binária - Implemente a busca binária para encontrar um elemento em uma lista ordenada.


```python
l = [1, 2, 3, 4, 5];
ne = 5; # numero escolhido

i = 0; # inicio
f = len(l) - 1; # final
achou = False; # começa sem achar o valor

while i <= f and not achou: # procura do começo ao fim sem ter achado o valor ainda
  m = (i + f) // 2; # meio // divide por dois pois precisa ser um numero inteiro

  if l[m] == ne: # se o valor digitado esta na lista
    achou = True;
    print();
    print(f'Número escolhido: {ne}');
    print(f'Índice: {m}');
  elif l[m] < ne: # valor do meio é menor entao esta a direita do meio
    i = m + 1; # descarta o meio e a esquerda
  else:
    f = m - 1; # descarta o meio e a direita

if not achou: # se o while nao achou
  print(f'Número {ne} não esta na lista.');
```

22. Criptografia de Cifra de César - Crie uma função para criptografar uma mensagem usando a cifra de César.


```python
# cifra de cesar = desloca letras num numero de posiçoes

def cc(m, d): # def = definir funçao // recebe mensagem e deslocamento
  r = ""; # strig vazia para acumular o texto criptografado

  for l in m:
    if l.isalpha(): # .isalpha() = ve se a é uma letra do alfabeto (nao simbolo, numero ou espaço)

      if l.isupper(): # base minuscula ou maiuscula
        b = ord('A'); # codigo da letra A
      else:
        b = ord('a'); # codigo da letra a

      cl = ord(l); # codigo numerico da letra
      p = cl - b; # transforma o codigo da letra em uma posiçao dentro do alfabeto
      pd = p + d; # determina a posiçao para deslocar
      pf = pd % 26; # nao passa de Z (25)

      nl = chr(pf + b); # pega o numero e devolve a letra correspondente
      r = r + nl;
    else: # caso nao for letra ele nao sofre deslocamento
      r = r + l;

  return r;

print('Criptografando mensagens')
print();
m = input('Digite a mensagem: ');
d = int(input('Digite o deslocamento'));

mc = cc(m, d);
print();
print(f'Mensagem original: {m}');
print(f'Mensagem criptografada: {mc}');
```

23. Descriptografar Cifra de César - Crie uma função para descriptografar uma mensagem cifrada com a cifra de César.


```python
# estou usando a mesma base de codigo do exercicio anterior

def cc(m, d):
  r = "";

  for l in m:
    if l.isalpha():

      if l.isupper():
        b = ord('A');
      else:
        b = ord('a');

      cl = ord(l);
      p = cl - b;
      pd = p - d; # unica mudança do código // para descriptografar vai desfazer o deslocamento
      pf = pd % 26;

      lo = chr(pf + b);
      r = r + lo;
    else:
      r = r + l;

  return r;

print('Descriptografando mensagens')
print();
mc = input('Mensagem criptografada: ');
d = int(input('Deslocamento:'));

md = cc(mc, d);
print();
print(f'Mensagem original: {md}');
print(f'Mensagem criptografada: {mc}');
```

24. Validação de CPF - Implemente uma função para validar um CPF brasileiro.


```python
cpf = input('Digite o cpf: ');

if len(cpf) != 11:
  print("CPF INVÁLIDO");
elif cpf == cpf[0] * 11: # repete o primeiro caractere 11 vezes pra ver se é igual ao cpf por inteiro
  print("CPF INVÁLIDO");
else:

# apenas os dois ultimos numeros do cpf sao validos para verificaçao real segundo pesquisas

  # calculo primeiro digito
  s1 = 0; # primeira soma
  p = 10; # peso começa com 10 e vai diminuindo

  i = 0;
  while i < 9:
    d = int(cpf[i]); # converte o caractere dessa posiçao para um numero
    s1 = s1 + d * p; # em loop multiplica o digito pelo peso atual e soma no total
    p = p - 1;
    i = i + 1;

  r1 = (s1 * 10) % 11; # formula = multiplica a soma por 10, o resto divide por 11 //
  if r1 == 10:
    r1 = 0;

  # calculo segundo digito
  s2 = 0; # segunda soma
  p = 11; # peso começa com 11 (pois é o segundo) e vai diminuindo

  i = 0;
  while i < 10: # mesma logica do calculo para o primeiro numero
    d = int(cpf[i]);
    s2 = s2 + d * p;
    p = p - 1;
    i = i + 1;

  r2 = (s2 * 10) % 11;
  if r2 == 10:
    r2 = 0;

  real1 = int(cpf[9]); # inidice 9 = decimo numero
  real2 = int(cpf[10]); # indice 10 = decimo primeiro numero (ultimo)

  if r1 == real1 and r2 == real2: # compara os valores com valores reais // and = compara ao mesmo tempo se sao reais
    print('CPF válido');
  else:
    print('CPF inválido');
```

25. Classificação de Números - Classifique uma lista de números em três categorias: positivos, negativos e zeros.


```python
l = [-2, -1, 0, 1, 2];

p = [];
n = [];
z = [];

for num in l:
  if num > 0: # lista positivos
    p.append(num);
  elif num < 0: # lista negativos
    n.append(num);
  else: # lista zero
    z.append(num);

print(f'Lista original: {l}');
print();
print(f'Positivos: {p}');
print(f'Negativos: {n}');
print(f'Zeros: {z}');
```
